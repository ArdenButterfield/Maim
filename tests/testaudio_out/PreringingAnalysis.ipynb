```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
from IPython.display import Audio
```

```python
def envelope_follow(x_array):
    envelope = x_array * 0
    state = 0
    for i, x in enumerate(x_array):
        state *= 0.998
        state = max(state, x)
        envelope[i] = state
    return envelope
```

```python
encs = ['blade', 'lame', 'opus']

ws = []
fs = 0
fs, raw = wavfile.read('../testaudio/two_hits.wav')
raw = raw.astype(np.float32) / max(np.abs(raw))
for e in encs:
    _, data = wavfile.read(f'two_hits_{e}_32.wav')
    ws.append(data.astype(np.float32) / max(abs(data)))

plt.plot(envelope_follow(raw))
for i in ws:
    plt.plot(envelope_follow(i))
plt.legend(['raw'] + encs)
```
